```python
# 19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).

import numpy as np

vetor1 = np.random.randint(1, 101, 20)
vetor2 = np.random.randint(1, 101, 20)

produto = np.dot(vetor1, vetor2)

norma1 = np.linalg.norm(vetor1)
norma2 = np.linalg.norm(vetor2)

cos_angulo = produto / (norma1 * norma2)

angulo = np.degrees(np.arccos(cos_angulo))

print("Ângulo:", angulo, "graus")
```

```python
# 1. Calcule a integral da funcao f(x) = x2 sin(x) no intervalo de 0 a π.

from scipy import integrate
import numpy as np

def f(x):
    return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)

print("Resultado:", resultado)
print("Erro estimado:", erro)
```

```python
# 2. Resolva a equacao diferencial dy dx = x− y, com y(0) = 1, no intervalo de 0 a 10.

from scipy.integrate import solve_ivp
import numpy as np

def equacao(x, y):
    return x - y

x = np.linspace(0, 10, 100)

solucao = solve_ivp(
    equacao,
    [0, 10],
    [1],
    t_eval=x
)

print("Valores de x:")
print(solucao.t)

print("Valores de y:")
print(solucao.y[0])
```

```python
# 3. Encontre as raızes da funcao f(x) = x4 − 3x3 + 2 utilizando SciPy.

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

solucao = root(f, 1)

print("Raiz encontrada:", solucao.x)
```

```python
# 4. Use a funcao de otimizacao de SciPy para maximizar f(x) =−x2 +4x+1.

from scipy.optimize import minimize

def funcao(x):
    return -(-x**2 + 4*x + 1)

resultado = minimize(funcao, 0)

print("Máximo encontrado em x =", resultado.x)
print("Valor máximo =", -resultado.fun)
```

```python
# 8. Resolva o sistema de equacoes 3x+ 2y= 5 e x+ 4y= 6.

import numpy as np
from scipy import linalg

A = np.array([
    [3, 2],
    [1, 4]
])

B = np.array([5, 6])

solucao = linalg.solve(A, B)

print("x =", solucao[0])
print("y =", solucao[1])
```

```python
# 9. Ajuste uma curva quadratica aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).

import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

parametros, _ = curve_fit(parabola, x, y)

print("a, b, c =", parametros)
```

```python
# 10. Calcule a integral dupla da funcao f(x,y) = xy+ x2 no intervalo [0,2] para x e [0,1] para y.

from scipy.integrate import dblquad

def f(y, x):
    return x * y + x**2

resultado, erro = dblquad(
    f,
    0, 2,
    lambda x: 0,
    lambda x: 1
)

print("Resultado:", resultado)
print("Erro:", erro)
```

```python
# 15. Calcule a transformada de Fourier inversa de um sinal complexo gerado aleatoriamente.

import numpy as np
from scipy.fft import fft, ifft

sinal = np.random.rand(20) + 1j * np.random.rand(20)

transformada = fft(sinal)

sinal_recuperado = ifft(transformada)

print("Sinal original:")
print(sinal)

print("Sinal recuperado:")
print(sinal_recuperado)
```

```python
# 16. Resolva a equacao diferencial de segunda ordem d2 y dx2 +y= 0, com condicoes iniciais y(0) = 1 e y′(0) = 0.

from scipy.integrate import solve_ivp
import numpy as np

def sistema(x, Y):
    y, derivada = Y

    return [derivada, -y]

x = np.linspace(0, 10, 100)

solucao = solve_ivp(
    sistema,
    [0, 10],
    [1, 0],
    t_eval=x
)

print("y(x):")
print(solucao.y[0])
```

```python
# 6. Plote dois graficos de linha em subplots diferentes: y1 = ex e y2 = log(x).

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.1, 10, 200)

y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("y = e^x")

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("y = log(x)")

plt.tight_layout()
plt.show()
```

```python
# 8. Plote um grafico de linha com intervalos de confianca (sombramento) para a funcao y= sin(x).

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)

y = np.sin(x)

limite_superior = y + 0.1
limite_inferior = y - 0.1

plt.plot(x, y, label="sin(x)")

plt.fill_between(
    x,
    limite_inferior,
    limite_superior,
    alpha=0.3,
    label="Intervalo"
)

plt.title("Função Seno com Intervalo")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python
# 10. Plote tres graficos de linha em um unico grafico, cada um com uma cor e estilo de linha diferentes.

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), color="blue", linestyle="-")
plt.plot(x, np.cos(x), color="red", linestyle="--")
plt.plot(x, np.sin(2 * x), color="green", linestyle=":")

plt.title("Três Funções")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
# 11. Crie um grafico de barras empilhadas para tres categorias com tres grupos de valores.


import matplotlib.pyplot as plt

categorias = ["A", "B", "C"]

grupo1 = [3, 5, 4]
grupo2 = [2, 4, 3]
grupo3 = [1, 2, 5]

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")

base = [grupo1[i] + grupo2[i] for i in range(3)]

plt.bar(categorias, grupo3, bottom=base, label="Grupo 3")

plt.title("Barras Empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.legend()

plt.show()
```

```python
# 13. Gere um grafico de dispersao 3D para 100 pontos aleatorios utilizando Matplotlib

import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

ax.scatter(x, y, z)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
# 18. Crie um grafico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D"]
valores = [10, -5, 15, -8]

cores = [
    "green" if valor >= 0 else "red"
    for valor in valores
]

plt.bar(categorias, valores, color=cores)

plt.axhline(0)

plt.title("Valores Positivos e Negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")

plt.show()
```

```python
# 19. Plote um grafico de linha com duas series e adicione uma legenda para diferencia-las.

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="Seno")
plt.plot(x, y2, label="Cosseno")

plt.title("Duas Séries")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()

plt.show()
```

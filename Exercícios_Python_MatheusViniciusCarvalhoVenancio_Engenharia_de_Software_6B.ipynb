```python
# 21. Busca Binária

lista = [1,3,5,7,9,11,13,15]

numero = int(input('Digite o numero que deseja buscar: '))
inicio = 0
fim = len(lista) - 1

econtrou = False

while inicio <=fim:
  meio = (inicio + fim) // 2

  if lista[meio] == numero:
    encontrou = True
    break

  elif numero> lista[meio]:
    inicio = meio + 1
  else:
    fim = meio - 1

if encontrou:
  print('Número encontrado')
else:
  print('Número nao encotrado')
```

```python
#30. Jogo da Forca

palavra = 'python'

tentativas = 6
acertos = []

while tentativas > 0:

    palavra_mostrada = ''

    for letra in palavra:

        if letra in acertos:
            palavra_mostrada += letra
        else:
            palavra_mostrada += '_'

    print('Palavra:', palavra_mostrada)

    if '_' not in palavra_mostrada:
        print('Você ganhou!')
        break

    letra = input('Digite uma letra: ')

    if letra in palavra:
        acertos.append(letra)
        print('Você acertou!')

    else:
        tentativas -= 1
        print('Você errou!')
        print('Tentativas restantes:', tentativas)

if tentativas == 0:
    print('Você perdeu!')
    print('A palavra era:', palavra)
```

```python
#31. Ordenação por Bolha

lista = [5, 2, 8, 1, 3]

for i in range(len(lista)):

    for j in range(len(lista) - 1):

        if lista[j] > lista[j + 1]:

            auxiliar = lista[j]

            lista[j] = lista[j + 1]

            lista[j + 1] = auxiliar

print('Lista ordenada:', lista)
```

```python
#46. Calcular Média e Desvio Padrão

import math

lista = [10, 20, 30, 40, 50]

soma = 0

for numero in lista:

    soma += numero

media = soma / len(lista)

soma_desvio = 0

for numero in lista:

    soma_desvio += (numero - media) ** 2

variancia = soma_desvio / len(lista)

desvio = math.sqrt(variancia)

print('Média:', media)
print('Desvio padrão:', desvio)
```

```python
#47. Contar Números em Lista

lista = [1, 2, 2, 3, 3, 3, 4, 4]

numeros_contados = []

for numero in lista:

    if numero not in numeros_contados:

        contador = 0

        for valor in lista:

            if valor == numero:
                contador += 1

        print(numero, 'aparece', contador, 'vez(es)')

        numeros_contados.append(numero)
```

```python
#48. Ordenação por Inserção

lista = [5, 2, 4, 1, 3]

for i in range(1, len(lista)):

    numero = lista[i]

    j = i - 1

    while j >= 0 and lista[j] > numero:

        lista[j + 1] = lista[j]

        j -= 1

    lista[j + 1] = numero

print('Lista ordenada:', lista)
```

```python
#49. Soma dos Dígitos

numero = int(input('Digite um número inteiro: '))

numero = abs(numero)

soma = 0

while numero > 0:

    digito = numero % 10

    soma += digito

    numero = numero // 10

print('Soma dos dígitos:', soma)
```
